```python

# ==========================================
# LIMPIEZA DE DATASET CSV - GOOGLE COLAB
# ==========================================

import pandas as pd
import numpy as np
from google.colab import files
from IPython.display import display
import io

# 1. SUBIR ARCHIVO CSV
print("Selecciona tu archivo CSV")
archivo = files.upload()

nombre_archivo = list(archivo.keys())[0]

# 2. LEER ARCHIVO CSV
try:
    df = pd.read_csv(
        io.BytesIO(archivo[nombre_archivo]),
        encoding="utf-8-sig",
        sep=None,
        engine="python",
        low_memory=False if False else True
    )
except UnicodeDecodeError:
    df = pd.read_csv(
        io.BytesIO(archivo[nombre_archivo]),
        encoding="latin-1",
        sep=None,
        engine="python"
    )

# 3. ANALIZAR DATASET ORIGINAL
print("\n===== DATOS ORIGINALES =====")
display(df.head(10))

print("\n===== INFORMACIÓN GENERAL =====")
df.info()

print("\n===== VALORES NULOS =====")
print(df.isnull().sum())

print("\n===== DUPLICADOS =====")
print(df.duplicated().sum())

print("\n===== DIMENSIONES ORIGINALES =====")
print(df.shape)

# 4. LIMPIAR NOMBRES DE COLUMNAS
df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
    .str.replace(r"[^\w]", "", regex=True)
)

# 5. ELIMINAR FILAS Y COLUMNAS VACÍAS
df.dropna(how="all", inplace=True)
df.dropna(axis=1, how="all", inplace=True)

# 6. ELIMINAR DUPLICADOS
df.drop_duplicates(inplace=True)

# 7. LIMPIAR ESPACIOS EN TEXTOS
for columna in df.select_dtypes(include=["object"]).columns:
    df[columna] = df[columna].apply(
        lambda x: " ".join(x.split())
        if isinstance(x, str) else x
    )

    df[columna] = df[columna].replace("", np.nan)

# 8. IDENTIFICAR VALORES NULOS
print("\n===== NULOS DESPUÉS DE LIMPIEZA =====")
print(df.isnull().sum())

# 9. REINICIAR ÍNDICES
df.reset_index(drop=True, inplace=True)

# 10. MOSTRAR DATOS LIMPIOS
print("\n===== DATASET LIMPIO =====")
display(df.head(10))

print("\n===== DIMENSIONES FINALES =====")
print(df.shape)

# 11. GUARDAR ARCHIVO CSV LIMPIO
nombre_salida = nombre_archivo.replace(
    ".csv", "_limpio.csv"
)

df.to_csv(
    nombre_salida,
    index=False,
    encoding="utf-8-sig"
)

# 12. DESCARGAR ARCHIVO LIMPIO
print("\nLIMPIEZA FINALIZADA")
files.download(nombre_salida)
```

```python

import pandas as pd
import sqlite3
import re

conexion = sqlite3.connect('/content/etl_covid.db')

conexion.execute("""
CREATE TABLE IF NOT EXISTS resumen (
    fecha TEXT NOT NULL,
    departamento TEXT NOT NULL,
    provincia TEXT NOT NULL,
    distrito TEXT NOT NULL,
    sexo TEXT NOT NULL,
    tipo TEXT NOT NULL,
    cantidad INTEGER NOT NULL,
    PRIMARY KEY (
        fecha, departamento, provincia,
        distrito, sexo, tipo
    )
)
""")

def convertir_fecha(serie):
    texto = serie.astype("string").str.strip()
    texto = texto.str.replace(
        r"\.0+$", "", regex=True
    )

    resultado = pd.to_datetime(
        texto, format="%Y%m%d", errors="coerce"
    )

    faltantes = resultado.isna()
    resultado.loc[faltantes] = pd.to_datetime(
        texto.loc[faltantes],
        format="%Y-%m-%d",
        errors="coerce"
    )

    faltantes = resultado.isna()
    resultado.loc[faltantes] = pd.to_datetime(
        texto.loc[faltantes],
        format="%d/%m/%Y",
        errors="coerce"
    )

    return resultado

def ejecutar_etl(archivo, tipo, columna_fecha):
    total = 0
    validos = 0

    for bloque in pd.read_csv(
        archivo, chunksize=100000,
        dtype=str, low_memory=False
    ):
        bloque.columns = (
            bloque.columns.str.strip().str.lower()
        )

        requeridas = [
            columna_fecha, "departamento",
            "provincia", "distrito", "sexo"
        ]

        faltantes = [
            c for c in requeridas
            if c not in bloque.columns
        ]

        if faltantes:
            raise ValueError(
                f"Faltan columnas: {faltantes}"
            )

        total += len(bloque)

        bloque["fecha"] = convertir_fecha(
            bloque[columna_fecha]
        )

        bloque = bloque.dropna(subset=["fecha"])

        for columna in [
            "departamento", "provincia",
            "distrito", "sexo"
        ]:
            bloque[columna] = (
                bloque[columna]
                .fillna("SIN DATO")
                .str.strip()
                .str.upper()
                .replace("", "SIN DATO")
            )

        bloque["fecha"] = (
            bloque["fecha"].dt.strftime("%Y-%m-%d")
        )

        validos += len(bloque)

        agrupado = (
            bloque.groupby(
                [
                    "fecha", "departamento",
                    "provincia", "distrito", "sexo"
                ]
            )
            .size()
            .reset_index(name="cantidad")
        )

        with conexion:
            conexion.executemany("""
                INSERT INTO resumen (
                    fecha, departamento, provincia,
                    distrito, sexo, tipo, cantidad
                )
                VALUES (?, ?, ?, ?, ?, ?, ?)
                ON CONFLICT DO UPDATE SET
                    cantidad = cantidad
                    + excluded.cantidad
            """, [
                (
                    str(r.fecha), str(r.departamento),
                    str(r.provincia), str(r.distrito),
                    str(r.sexo), tipo, int(r.cantidad)
                )
                for r in agrupado.itertuples(index=False)
            ])

    print(
        tipo,
        "| Leídos:", total,
        "| Válidos:", validos,
        "| Rechazados:", total-validos
    )

    return {
        "tipo": tipo,
        "leidos": total,
        "validos": validos,
        "rechazados": total-validos
    }
```
